```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import Normalizer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/ConcreteData_cleaned.csv")
```

```python
df.head()
```

There are no categorical variables in this data set so we will go straight to scaling the numerical columns. The first seven columns contain amounts of ingredients for concrete. They all have the same units, kg/m^3, however the scale varies quite a bit between features/ingredients. I'm not sure if it is better to scale each ingredient individually or, since each ingredient is a proportion of the whole mixture, we could normalize each record (row) independently of the others, such that all the component ingredients sum to 1. We will try both approaches. The Age column should not be included in the normalization and will be scaled separately.

```python
#Separate features and target variable
X = df.drop(columns=["Concrete compressive strength"])
y = df["Concrete compressive strength"]
```

```python
#Train/test split
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.8, random_state=10)
```

# Normalized per row:

```python
transformer = Normalizer(norm='l1')
transformer.fit(X_train.drop(columns=['Age']))
X_train_norm = transformer.transform(X_train.drop(columns=['Age']))
```

```python
Scaler = MinMaxScaler()
Scaler.fit(X_train[['Age']])
X_train_Age_scaled = Scaler.transform(X_train[['Age']])
```

```python
X_test_norm = transformer.transform(X_test.drop(columns=['Age']))
X_test_Age_scaled = Scaler.transform(X_test[['Age']])
```

```python
X_train_normalized = X_train.copy()
X_train_normalized.iloc[:,0:7] = X_train_norm
X_train_normalized['Age'] = X_train_normalized['Age'].astype(float)
X_train_normalized.loc[:,'Age'] = X_train_Age_scaled
```

```python
X_train_normalized.head()
```

```python
X_test_normalized = X_test.copy()
X_test_normalized.iloc[:,0:7] = X_test_norm
X_test_normalized['Age'] = X_test_normalized['Age'].astype(float)
X_test_normalized.loc[:,'Age'] = X_test_Age_scaled
```

```python
X_test_normalized.head()
```

```python
X_train_normalized.to_csv("data/X_train_normalized.csv")
X_test_normalized.to_csv("data/X_test_normalized.csv")
```

# Scaled per column:

```python
Scaler.fit(X_train)
X_train_scale = Scaler.transform(X_train)
X_test_scale = Scaler.transform(X_test)
```

```python
X_train_scaled = X_train.copy()
X_train_scaled['Age'] = X_train_scaled['Age'].astype(float)
X_train_scaled.loc[:,:] = X_train_scale
X_test_scaled = X_test.copy()
X_test_scaled['Age'] = X_test_scaled['Age'].astype(float)
X_test_scaled.loc[:,:] = X_test_scale
```

```python
X_train_scaled.to_csv("data/X_train_scaled.csv")
X_test_scaled.to_csv("data/X_test_scaled.csv")
```
